```python
# 5. Selecting numerical , categorical features
```

```python
numerical_col = x.select_dtypes(include = ['int64','float64']).columns.tolist()
categorical_col = x.select_dtypes(include = ['object','category']).columns.tolist()
```

```python
# 6. numerical preprocessing with 'Pipeline '
```

```python
numerical_pipeline = Pipeline([
                             ('imputer',SimpleImputer(strategy = "median")),
                             ('scaler',StandardScaler())
                              ])
                              
```

```python
# 7. Categorical preprocessing
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
])
```

```python
# 8. combiend preprcoessing
```

```python
preprocessor = ColumnTransformer([
    ('numerical',numerical_pipeline,numerical_col),
    ('categorical', categorical_pipeline, categorical_col)
])
```

```python
# 9 fit_transform
```

```python
x_clean = preprocessor.fit_transform(data)
```

```python
    # 10 columns names
```

```python
col_name = preprocessor.get_feature_names_out()
```

```python
x_clean = pd.DataFrame(
    x_clean,
    columns = col_name
)
```

```python
# 11. Check final dataset
print("\nCleaned Dataset:")
print(x_clean.head())

print("\nShape of final dataset:")
print(x_clean.shape)

print("\nMissing values AFTER preprocessing:")
print(x_clean.isnull().sum().sum())
```
