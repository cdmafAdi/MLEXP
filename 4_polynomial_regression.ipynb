```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.linear_model import LinearRegression
# from sklearn
```

```python
data = sns.load_dataset("mpg")
```

```python
data = pd.DataFrame(data)
data.info()
```

```python
data = data.dropna(subset=["horsepower"])
```

```python
x = data.drop(columns=['mpg','origin','name'])
y = data['mpg']
x
```

```python
y
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,random_state=42)
```

```python
x_train.shape
y_train.shape
```

```python
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
train_err = []
test_err = []

for deg in range(1,11):

    poly = PolynomialFeatures(degree=deg)

    x_train_poly = poly.fit_transform(x_train_scaled)
    x_test_poly = poly.transform(x_test_scaled)

    model1 = LinearRegression()

    model1.fit(x_train_poly,y_train)
    train_pred = model1.predict(x_train_poly)
    test_pred = model1.predict(x_test_poly
                               )

    train_mse = mean_squared_error(y_train,train_pred)
    # test_mse = mean_squared_error(y_test,x_test_poly)

    train_err.append(train_mse).mean
    # test_err.append(test_mse)
    print(
        "Degree:", deg,
        "Train MSE:", round(train_err, 2),
        # "Test MSE:", round(test_err, 2)
    )

```
