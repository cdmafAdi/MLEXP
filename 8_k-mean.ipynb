```python
# Experiment 8: K-Means Clustering on Iris Dataset

# 1. Import libraries
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
# 2. Load Iris dataset
iris = load_iris()

# Take only the features
X = iris.data

# Do NOT use iris.target because clustering should be unsupervised
```

```python

# 3. Find the best K using Elbow Method
inertia = []

for k in range(2, 8):
    model = KMeans(n_clusters=k, n_init=10, random_state=42)
    model.fit(X)
    inertia.append(model.inertia_)

# Plot Elbow graph
plt.plot(range(2, 8), inertia, marker='o')
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.show()
```

```python


# 4. Find the best K using Silhouette Score
silhouette_scores = []

for k in range(2, 8):
    model = KMeans(n_clusters=k, n_init=10, random_state=42)
    labels = model.fit_predict(X)

    score = silhouette_score(X, labels)
    silhouette_scores.append(score)

# Plot Silhouette scores
plt.plot(range(2, 8), silhouette_scores, marker='o')
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Method")
plt.show()
```

```python



# 5. Display silhouette scores
for k, score in zip(range(2, 8), silhouette_scores):
    print("K =", k, "Silhouette Score =", round(score, 3))


# 6. Apply K-Means with K = 3
kmeans = KMeans(n_clusters=3, n_init=10, random_state=42)
clusters = kmeans.fit_predict(X)

# 7. Visualize the clusters
plt.scatter(X[:, 0], X[:, 1], c=clusters)
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("K-Means Clustering (K = 3)")
plt.show()
```
