# 0 . improt libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_diabetes
from sklearn.metrics import mean_squared_error, r2_score, accuracy_score
from sklearn.model_selection import train_test_split
```

```python
from sklearn.linear_model import LinearRegression as LR
```

# 1, import data

```python
data = load_diabetes()

x = data.data
y = data.target
```

```python
import pandas as pd
from sklearn.datasets import load_diabetes

diabetes = load_diabetes()

data = pd.DataFrame(
    diabetes.data,
    columns=diabetes.feature_names
)

data['target'] = diabetes.target

data.head()
```

```python
# 3. train test split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, random_state=42, test_size=0.3)
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x_train)

x_test = scaler.transform(x_test)
```

```python
# 4. lienar regressionn 
```

```python
model = LR()
model.fit(x_train,y_train)
model_pred = model.predict(x_test)

```

```python
# evaluation matrix
```

```python
mse = mean_squared_error(y_test,model_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test,model_pred)
```

```python
print("mse = ", mse)
print("rmse = ",rmse)
print("r2 =",r2)
```

```python
model2 = LR()
model2.fit(x_scaled,y_train)
model2_pred = model2.predict(x_test)
```

```python
mse = mean_squared_error(y_test,model2_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test,model2_pred)
```

```python
print("mse = ", mse)
print("rmse = ",rmse)
print("r2 =",r2)
```

```python
# Step 10: Clear Actual vs Predicted plot

plt.figure(figsize=(8, 6))

# Plot actual and predicted values
plt.scatter(y_test, model2_pred, alpha=0.7)

# Ideal prediction line
# If prediction is perfect, predicted = actual
plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    linestyle="--"
)

plt.xlabel("Actual Diabetes Progression")
plt.ylabel("Predicted Diabetes Progression")
plt.title("Actual vs Predicted Values")

plt.grid(True, alpha=0.3)
plt.show()
```
