```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix, accuracy_score
import matplotlib.pyplot as plt
```

```python
data = load_wine()

x = data.data
y = data.target
```

```python
x.shape
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y, random_state=42,test_size=0.2)
```

```python
scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
linear_svm = SVC(kernel = 'linear')
poly_svm = SVC(kernel = 'poly')
rbf_svm = SVC(kernel ="rbf")
```

```python
linear_svm.fit(x_train_scaled,y_train)
l_pred=linear_svm.predict(x_test_scaled)
```

```python
poly_svm.fit(x_train_scaled,y_train)
p_pred=poly_svm.predict(x_test_scaled)
```

```python
rbf_svm.fit(x_train_scaled,y_train)
r_pred=rbf_svm.predict(x_test_scaled)
```

```python
linear_acc = accuracy_score(y_test,l_pred)
linear_cm = confusion_matrix(y_test,l_pred)
poly_acc = accuracy_score(y_test,p_pred)
poly_cm = confusion_matrix(y_test,p_pred)
rbf_acc = accuracy_score(y_test,r_pred)
rbf_cm = confusion_matrix(y_test,r_pred)
```

```python
print("linear_svm acc:",linear_acc)
print(linear_cm)
print("poly_svm acc:",poly_acc)
print(poly_cm)
print("rbf_svm acc:",rbf_acc)
print(rbf_cm)
```

```python
kernels = ['linear','poly','rbf']
accuracies = [linear_acc,poly_acc,rbf_acc]

plt.bar(kernels,accuracies)
plt.title("SVM Kernel Accuracies")
plt.xlabel("Kernel")
plt.ylabel("Accuracy")
plt.show()
```
